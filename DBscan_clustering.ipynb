## DBSCAN CLUSTERING (DENSITY BASED)

### Load the data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans, DBSCAN
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv(r'Mall_customers.csv')
df
```

### Feature selection

- To put in arrays add .values

```python
x = df[['Annual Income (k$)','Spending Score (1-100)']].values
x
```

### Visualization

```python
# the :,0 bring the first column in x axis and :,1 brings the other in y axis
plt.scatter(x[:,0],x[:,1])
```

### Scaling

```python
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)
x_scaled
```

### DBScan
- epsilon and min_samples define dbscan

```python
dbscan = DBSCAN(eps=0.5, min_samples=5)
labels = dbscan.fit_predict(x_scaled)
```

```python
labels
```

```python
unique_labels = np.unique(labels)
```

```python
unique_labels
```

```python
for label in unique_labels:
    if label == -1:
        plt.scatter(x[labels==label,0],
        x[labels==label,1],
        color = 'black', label = 'outlier')
    else:
        plt.scatter(x[labels==label,0],
        x[labels==label,1],
        color = 'orange', label =f'Clusters {label}')
        plt.legend()
```

### Metric scores
- more than 0.5 mean its a strong cluster
- b2n 0.25-0.5 mean its reasonable 
- less than 0.25 means weak

```python
from sklearn.metrics import silhouette_score
score = silhouette_score(x_scaled, labels)
score
```

### DB_Score
- less than 0.5 means excellent model
- close to 1 is acceptable
- greater than 2 is poor model

```python
from sklearn.metrics import davies_bouldin_score
db_score = davies_bouldin_score(x_scaled, labels)
db_score
```
